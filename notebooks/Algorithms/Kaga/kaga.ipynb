```python
# ---- 使用分析/对比图工具 ----
import numpy
from lowlight_image_enhance import utils, colorspace, algorithms
from lowlight_image_enhance.analysis import ana_light
import matplotlib.pyplot as plt

# 1. 读取并转换到 LCH，取 L 层data\lol\low\44.png
low_bgr_u8 = utils.imread("../../../data/lol/low/44.png")
ref_bgr_u8 = utils.imread("../../../data/lol/high/44.png")
low_l = colorspace.BGR2LCH(utils.normalize(low_bgr_u8))[:, :, 0]
ref_l = colorspace.BGR2LCH(utils.normalize(ref_bgr_u8))[:, :, 0]

# 2. 几种算法的增益结果
kaga = algorithms.curves.kaga(low_l)
kaga_autok = algorithms.curves.kaga_autok(low_l)


# 3. 调用工具画对比图（参照 + 算法各一行）
ana_light.plot_compare(low_l, ref=ref_l, kaga=kaga, kaga_autok=kaga_autok)
plt.show()
```

```python
import matplotlib.pyplot as plt
from lowlight_image_enhance import utils, colorspace, algorithms

# ---- 3x2 网格：六张灰度图并排对比 ----

# 1. 准备数据：dict 的插入顺序 = 网格的填充顺序（3 行 2 列，按行从左到右）
low_l = colorspace.BGR2LCH(utils.normalize(utils.imread("low/44.png")))[:, :, 0]
ref_l = colorspace.BGR2LCH(utils.normalize(utils.imread("high/44.png")))[:, :, 0]

panels = {
    "low (input)":                  low_l,
    "ref":                          ref_l,
    "gamma 0.35":                   algorithms.curves.gamma(low_l, 0.35),
    "k*L^gamma":                    algorithms.curves.kaga_k(low_l),
    "k*(L+a)^gamma":                algorithms.curves.kaga_ka(low_l),
    "L*(k*l^(gamma-1)).clip(0,50)": algorithms.curves.kaga(low_l, upper_limit=50),
}

# 2. 建画布：3 行 2 列
fig, axes = plt.subplots(2, 3, figsize=(18, 9))

# 3. 填充：axes.flat 把 3x2 的二维数组拍成一维，可以直接 for + zip
for ax, (name, L) in zip(axes.flat, panels.items()):
    ax.imshow(L, cmap="gray", vmin=0.0, vmax=1.0)  # 固定值域，六张图才能公平比亮度
    ax.set_title(name)
    ax.axis("off")                                 # 灰度图不需要坐标轴刻度

# 4. 收紧留白，防止标题/子图互相压住
fig.tight_layout()
plt.show()
```
